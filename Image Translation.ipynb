# **Image Translation**

Write a Python program using OpenCV to perform image translation on the **"Given image"**.

The program should:

1. Import `cv2`, `numpy`, and `matplotlib.pyplot`.
2. Read the image **"given** in color mode.
3. Create a **translation matrix** that moves the image  
   **50 pixels to the right** and **80 pixels down**.
4. Apply the translation
5. Display the following using a **1 × 2 layout** in Matplotlib:  
   - **Original Image**  
   - **Translated Image**


```python
# Import 

import cv2
import numpy as np
import matplotlib.pyplot as plt


```

```python
# Read the image using OpenCV
img = cv2.imread('baseball.jpg', cv2.IMREAD_COLOR)



```

```python
# Display the image using Matplotlib
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis('off')
plt.show()


```

```python

# Get image dimensions
rows, cols = img.shape[:2]

# Create the translation matrix
M = np.float32([[1, 0, 50],
                [0, 1, 80]])

# Apply translation
translated_img = cv2.warpAffine(img, M, (cols, rows))

```

```python

plt.imshow(cv2.cvtColor(translated_img, cv2.COLOR_BGR2RGB))
plt.title("Translated Image")
plt.axis('off')
plt.show()

```

# **Image Scaling**

Write a Python program using OpenCV to perform image scaling on the **given image**.

The program should:

1. Scale the image to:
   - **0.5 times** its original size (downscaling)
   - **2 times** its original size (upscaling)
2. Use `cv2.resize` with appropriate scaling factors.
5. Display the following using a **1 × 3 layout** in Matplotlib:  
   - **Original Image**  
   - **Downscaled Image (0.5×)**  
   - **Upscaled Image (2×)**


```python

import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read the image
img = cv2.imread('baseball.jpg', cv2.IMREAD_COLOR)

# Downscale the image (0.5x)
downscaled_img = cv2.resize(img, None, fx=0.5, fy=0.5)

# Upscale the image (2x)
upscaled_img = cv2.resize(img, None, fx=2, fy=2)

# Display all three images in a 1x3 layout
plt.figure(figsize=(15, 5))

# Original Image
plt.subplot(1, 3, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis('off')

# Downscaled Image
plt.subplot(1, 3, 2)
plt.imshow(cv2.cvtColor(downscaled_img, cv2.COLOR_BGR2RGB))
plt.title("Downscaled Image (0.5x)")
plt.axis('off')

# Upscaled Image
plt.subplot(1, 3, 3)
plt.imshow(cv2.cvtColor(upscaled_img, cv2.COLOR_BGR2RGB))
plt.title("Upscaled Image (2x)")
plt.axis('off')

plt.tight_layout()
plt.show()
```

# **Image Shearing**

Write a Python program using OpenCV to perform image shearing on the **given image**.

The program should:

1. Create a **shearing transformation matrix** to apply:
   - **Horizontal shearing**  
   - **Vertical shearing**
2. Apply the shearing transformations.
3. Display the following using a **1 × 3 layout** in Matplotlib:  
   - **Original Image**  
   - **Horizontally Sheared Image**  
   - **Vertically Sheared Image**


```python
rows, cols = img.shape[:2]

M1 = np.float32([[1, 0.5, 0], [0, 1, 0]])
horizontal = cv2.warpAffine(img, M1, (cols, rows))

M2 = np.float32([[1, 0, 0], [0.5, 1, 0]])
vertical = cv2.warpAffine(img, M2, (cols, rows))
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(cv2.cvtColor(horizontal, cv2.COLOR_BGR2RGB))
plt.title("Horizontal Shearing")
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(cv2.cvtColor(vertical, cv2.COLOR_BGR2RGB))
plt.title("Vertical Shearing")
plt.axis('off')

plt.tight_layout()
plt.show()
```

# **Image Reflection**

Write a Python program using OpenCV to perform image reflection on the **given image**.

The program should:

1. Generate the following reflected versions of the image:
   - **Horizontal Reflection** (flip left ↔ right)  
   - **Vertical Reflection** (flip top ↔ bottom)  
   - **Both-axis Reflection**  
2. Use `cv2.flip` with appropriate flip codes.
3. Display the following using a **2 × 2 layout** in Matplotlib:  
   - **Original Image**  
   - **Horizontally Reflected Image**  
   - **Vertically Reflected Image**

```python
horizontal = cv2.flip(img, 1)
vertical = cv2.flip(img, 0)
both = cv2.flip(img, -1)
```

```python
plt.figure(figsize=(10, 8))

plt.subplot(2, 2, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis('off')

plt.subplot(2, 2, 2)
plt.imshow(cv2.cvtColor(horizontal, cv2.COLOR_BGR2RGB))
plt.title("Horizontal Reflection")
plt.axis('off')

plt.subplot(2, 2, 3)
plt.imshow(cv2.cvtColor(vertical, cv2.COLOR_BGR2RGB))
plt.title("Vertical Reflection")
plt.axis('off')

plt.subplot(2, 2, 4)
plt.imshow(cv2.cvtColor(both, cv2.COLOR_BGR2RGB))
plt.title("Both-axis Reflection")
plt.axis('off')

plt.tight_layout()
plt.show()
```

# **Image Rotation**

Write a Python program using OpenCV to perform **image rotation** on the **given image**.

The program should:

1. Rotate the image by:  
   - **45 degrees**  
   - **90 degrees**
2. Use `cv2.getRotationMatrix2D` and `cv2.warpAffine`.
3. Display the following using a **1 × 3 layout** in Matplotlib:  
   - **Original Image**  
   - **45° Rotated Image**  
   - **90° Rotated Image**


```python
rows, cols = img.shape[:2]
center = (cols // 2, rows // 2)

M1 = cv2.getRotationMatrix2D(center, 45, 1)
rotated45 = cv2.warpAffine(img, M1, (cols, rows))

M2 = cv2.getRotationMatrix2D(center, 90, 1)
rotated90 = cv2.warpAffine(img, M2, (cols, rows))
```

```python
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(cv2.cvtColor(rotated45, cv2.COLOR_BGR2RGB))
plt.title("45° Rotated Image")
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(cv2.cvtColor(rotated90, cv2.COLOR_BGR2RGB))
plt.title("90° Rotated Image")
plt.axis('off')

plt.tight_layout()
plt.show()
```
